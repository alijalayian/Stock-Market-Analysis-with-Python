```python
import yfinance as yf
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
tickers = [ "AAPL","MSFT","TSLA"]
data = yf.download(tickers, start = "2022-01-01",end = "2025-01-01",auto_adjust=True)
```

```python
print(data.head())
print(data.shape)
```

```python
close_price = data["Close"]
print (close_price.head())
print (close_price.describe())
```

```python
benchmark= yf.download("^GSPC", start = "2022-01-01",end = "2025-01-01")
benchmark_close = benchmark["Close"]
benchmark_returns= benchmark_close.pct_change().dropna()
benchmark_total_return = (benchmark_close.iloc[-1]/benchmark_close.iloc[0]-1)*100
print(f"S&P 500 total return : {benchmark_total_return}" )
```

```python
close_price["AAPL"].plot(figsize=(10,5),title="AAPL PRICE 2022-2025")
plt.xlabel("DATE")
plt.ylabel("PRICE")
plt.savefig('images/price_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
close_price.plot(figsize=(12,6), title="Stock Price Comparison")
plt.xlabel("Date")
plt.ylabel("Price (USD)")
plt.legend(tickers)
plt.savefig('images/price_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
daily_returns= close_price.pct_change()
print(daily_returns.head())
print(daily_returns.describe())
```

```python
volatility= daily_returns.std()
print (volatility)
```

```python
annual_volatility= volatility*(252**0.5)
print(annual_volatility)
```

```python
correlation_matrix = daily_returns.corr()
print(correlation_matrix)
```

```python
sns.heatmap(correlation_matrix, annot=True , cmap = "coolwarm")
plt.title("stock correlation matrix")
plt.savefig('images/correlation_heatmap.png', dpi=150, bbox_inches='tight')
plt.show()
```

```python
total_return = (close_price.iloc[-1]/close_price.iloc[0]-1)*100
print(total_return)
```

```python
summary= pd.DataFrame({
    'Total return %' : total_return,
    ' annual volatility %' : annual_volatility*100
})
print (summary)
```

```python

risk_free_rate = 0.04


annual_return = daily_returns.mean() * 252


sharpe_ratio = (annual_return - risk_free_rate) / annual_volatility
print(sharpe_ratio)
```

```python
summary['Sharpe Ratio'] = sharpe_ratio
print(summary)
```

```python
comparison = pd.DataFrame({
    'AAPL': close_price['AAPL'] / close_price['AAPL'].iloc[0] * 100,
    'MSFT': close_price['MSFT'] / close_price['MSFT'].iloc[0] * 100,
    'TSLA': close_price['TSLA'] / close_price['TSLA'].iloc[0] * 100,
    'S&P 500': benchmark_close.squeeze() / benchmark_close.iloc[0].squeeze() * 100
})

comparison.plot(figsize=(12,6), title="Normalized Performance vs S&P 500")
plt.xlabel("Date")
plt.ylabel("Normalized Price (Base=100)")
plt.savefig('images/benchmark_comparison.png', dpi=150, bbox_inches='tight')
plt.show()
```
